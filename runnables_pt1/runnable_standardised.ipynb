To bring standardisation -> the technique used is called Runnable

```python
from abc import ABC, abstractmethod
#abc- abstract based classes
```

```python
class Runnable(ABC): #abstract class since its inheritting from Runnable.

  @abstractmethod #just decorator
  def invoke(input_data):
    pass
```

```python
import random

class NakliLLM(Runnable): #task specific runnables.

  def __init__(self):
    print('LLM created')
 
  def invoke(self, prompt): #need to mandatorily implement this
    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}


#   def predict(self, prompt):

#     response_list = [
#         'Delhi is the capital of India',
#         'IPL is a cricket league',
#         'AI stands for Artificial Intelligence'
#     ]

#     return {'response': random.choice(response_list)}
```

now using it, kitni bhi badi chain bna skte.

```python
class RunnableConnector(Runnable): #helper. - runnable primitives

  def __init__(self, runnable_list):
    self.runnable_list = runnable_list

  def invoke(self, input_data):

    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data) #llm.invoke(input_data)-> in input they get output of the previous step. except just first run mei you get input_data as {'length':'long', 'topic':'india'}

    return input_data
```
